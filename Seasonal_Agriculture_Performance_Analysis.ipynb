### 5.7 Unusual Value Screening

Numerical variables are screened for unusually high or low observations using the Interquartile Range (IQR) method. The purpose is to identify potential outliers for further investigation rather than automatically removing them.

```python
# Identify potential outliers using the IQR method

numeric_columns = df.select_dtypes(include="number").columns

outlier_summary = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = ((df[column] < lower_bound) |
                     (df[column] > upper_bound)).sum()

    outlier_summary.append({
        "Column": column,
        "Lower_Bound": round(lower_bound, 2),
        "Upper_Bound": round(upper_bound, 2),
        "Potential_Outliers": outlier_count
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    by="Potential_Outliers",
    ascending=False
)
```

### 5.8 Visual Outlier Analysis

Box plots are used to visually examine the distribution and potential extreme values of important agricultural performance and economic variables.

```python
# Visualize potential outliers in key variables

key_variables = [
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3"
]

for column in key_variables:
    plt.figure(figsize=(9, 4))
    sns.boxplot(x=df[column])
    plt.title(f"Outlier Analysis: {column}")
    plt.xlabel(column)
    plt.tight_layout()
    plt.show()
```

### 5.9 Outlier Interpretation

The box plots show that several agricultural and economic variables contain observations beyond the IQR-based whiskers. Water usage, for example, shows a strongly right-skewed distribution with several high-value observations.

These extreme observations are treated as potential outliers rather than errors. Since farms can differ substantially in area, crop type, production level, and resource requirements, unusually high values may represent legitimate agricultural conditions. Therefore, the observations are retained for further analysis unless additional evidence indicates a data-quality problem.

## 8. Multivariate Analysis

Multivariate analysis examines the combined relationship between three or more variables to identify patterns that may not be visible when variables are analyzed individually.

### 8.1 Crop–Season Interaction on Average Yield

**Question:** How does average crop yield vary across different seasons for different crops?

Analyzing crop and season together helps identify whether the performance of a crop changes across seasons and whether certain crops consistently achieve higher yields under different seasonal conditions.

### 8.1 Interpretation

The heatmap shows clear differences in average yield across crops and seasons.

- **Sugarcane** records the highest average yield in all three seasons, with **53.46 tonnes/ha in Kharif, 43.29 tonnes/ha in Rabi, and 38.42 tonnes/ha in Zaid**.
- For most crops, the average yield is highest during **Kharif** and decreases through Rabi and Zaid.
- **Maize** and **Rice** show relatively higher yields compared with most other crops, while **Pulses** records the lowest average yield across the three seasons.
- The results indicate that crop performance varies across seasons, showing the importance of considering **crop and season together** rather than analyzing them independently.
- These observations represent associations within the dataset and should not be interpreted as proof that season alone causes differences in crop yield.

### 8.2 Season–Irrigation Interaction on Average Yield

**Question:** How does the average yield vary across different irrigation methods and seasons?
